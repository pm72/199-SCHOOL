### შედარების ოპერატორები

__`bool` ტიპი: `True / False`__

შედარების ოპერატორებია: `>, >=, <, <=, == !=`

`==`  –  ობიექტების შედარება ტოლობაზე

`!=`  –  ობიექტები არ არის ტოლი

შედეგად ბრუნდება ლოგიკური (`bool`) ტიპის მნიშვნელობა: `True` ან `False` (ჭეშმარიტი / მცდარი)

````python
print("1 == 1 is", 1 == 1)                  # True
print("14 == 21 is", 14 == 21)              # False
print("35 < 47 is", 35 < 47)                # True
print("16 > 12 is", 16 > 12)                # True
print("13 >= 13 is", 13 >= 13)              # True
print("54 <= 117 is", 54 <= 117)            # True
print("17 > 17 is", 17 > 17)                # False
print("17 < 17 is", 17 < 17)                # False
print("0 != 0 is", 0 != 0)                  # False
print("True != False is", True != False)    # True
````

__ლოგიკური შედარების ოპერატორები: `and`, `or`, `not`__

`and`, `or`, `not` ოპერატორები აბრუნებენ ლოგიკური შედარების შედეგს `True / False`

არსებობს შეშმარიტობის ცხრილი

````python
'''
ლოგიკური „და“ and               „ლოგიკური გამრავლება“
True and True = True                   1 * 1 = 1
True and False = False                 1 * 0 = 0
False and True = False                 0 * 1 = 0
False and False = False                0 * 0 = 0


ლოგიკური „ან“ or                  „ლოგიკური შეკრება“
True or True = True                    1 + 1 = 1
True or False = True                   1 + 0 = 1
False or True = True                   0 + 1 = 1
False or False = False                 0 + 0 = 0


ლოგიკური „უარყოფა“ not
not True = False
not False = True
'''
````

````python
# ყველაფერი რაც ნულისგან, სიცარიელისგან განსხვავებულია არის True (ჭეშმარიტი).
# ნული, სიცარიელე არის False (მცდარი).

'''
15, -54, 2.69, -0.89 "String", [12, 36], {'one': '1'} = True
0, 0.0, "", [], {} = False
'''

print(bool(15), bool(-5), bool(2.1), bool([12, 36]), bool({'two': 2}))
print(bool(0), bool(0.0), bool([]), bool({}))
````

_3. შექმენით ორი სტრინგის ტიპის ცვლადი `st1` და `st2`, შეადარეთ ისინი `is`-ის გამოყენებით, თუ ემთხვევა გამოიტანეთ ტექსტი `Same object`, წინააღმდეგ შემთხვევაში – `Different object`_

სტრიქონული ობიექტების `is` ოპერატორით შედარებისას საქმე გვაქვს `python`-ის საინტერესო ფუნქციონალთან –  `string interning`.

როცა ამ კოდს გაუშვებთ, შედეგი იქნება:

```py
Input text 140712345678912
Input text 140712345678912

Same Object
```

თუმცა, შეიძლება განსხვავებული შედეგიც გვქონდეს: `Different objects` – `id`-ები განსხვავებულია.

__რატომ ხდება ეს?__

`Python`-ში მოკლე სტრინგები, რომლებიც შეიცავს მხოლოდ ასოებს, რიცხვებს და ქვედა ტირეებს, ავტომატურად განიცდის `interning`-ს. ეს ნიშნავს, რომ `Python` მეხსიერების ოპტიმიზაციისთვის იყენებს მხოლოდ ერთ ობიექტს იდენტური სტრინგების ყველა შემთხვევისთვის.

````python
# Interning ხდება
a = "hello"
b = "hello"
print(a is b)  # True

# Interning არ ხდება (გრძელი სტრინგი სპეციალური სიმბოლოებით)
c = "hello world with special chars!"
d = "hello world with special chars!"
print(c is d)  # False (შეიძლება)
````



---

მოკლედ, `is` ოპერატორი ამოწმებს ობიექტის იდენტურობას, ხოლო `==` ოპერატორი ამოწმებს მნიშვნელობის თანასწორობას. სტრინგების შედარებისთვის ყოველთვის უმჯობესია `==` გამოიყენოთ.

---

#### რა არის `string interning`?

`String Interning` არის მეხსიერების ოპტიმიზაციის ტექნიკა, სადაც იდენტური სტრინგები ინახება მხოლოდ ერთხელ მეხსიერებაში.

##### როგორ მუშაობს?

`Python`-ი ინახავს სპეციალურ `intern table`-ს (ინტერნირებული სტრინგების ცხრილს), სადაც თითოეული უნიკალური სტრინგი ინახება მხოლოდ ერთხელ. როცა ახალი სტრინგი იქმნება, `Python` ამოწმებს:

1. თუ ეს სტრინგი უკვე არსებობს `intern table`-ში – აბრუნებს მის რეფერენსს

2. თუ არ არსებობს – ქმნის ახალს და ამატებს `table`-ში

##### რა სტრინგები განიცდის `interning`-ს ავტომატურად?

```py
# ✅ Interning ხდება:
a = "hello"          # ასოები
b = "hello123"       # ასოები + რიცხვები  
c = "hello_world"    # ასოები + underscore
d = ""               # ცარიელი სტრინგი

# ❌ Interning არ ხდება:
e = "hello world"    # სივრცე
f = "hello!"         # სპეციალური სიმბოლოები
g = "very long string that exceeds certain length limits"
```
---

##### მაგალითი:

````python
# ავტომატური interning
x = "python"
y = "python" 
print(x is y)                 # True
print(id(x) == id(y), "\n")   # True

# მანუალური interning
import sys
a = "hello world!"
b = "hello world!"
print(a is b, "\n")           # False (ავტომატურად არ interning-დება)

# მანუალური interning
a = sys.intern("hello world!")
b = sys.intern("hello world!")
print(a is b, "\n")           # True
````



---

##### უპირატესობები:

* მეხსიერების დაზოგვა: იდენტური სტრინგები იყენებს ერთსა და იმავე მეხსიერების ადგილს

* სწრაფი შედარება: `is` ოპერატორით შედარება `O(1)` დროში

* `Dictionary key`-ების ოპტიმიზაცია: `hash`-ირება უფრო ეფექტურია

##### გაფრთხილება:

`String interning`-ზე არ უნდა დაეყრდნოთ კოდში! ყოველთვის გამოიყენეთ `==` სტრინგების შედარებისთვის და არა `is`.

````python
import sys

st1 = sys.intern(input("Enter first string: "))
st2 = sys.intern(input("Enter second string: "))

print("Same object" if st1 is st2 else "Different object")
````
